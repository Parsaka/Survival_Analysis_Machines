```r
# Load necessary libraries
library(survival)

# Define the adjusted baseline intensity function alpha(t)
alpha_t <- function(t) {
  return(1 + sin(2 * pi * t / 365))  # Shift sine function upward by 1
}

# Define the cumulative hazard function H0(t) using integrate
H0_t <- function(t) {
  sapply(t, function(x) {
    integrate(alpha_t, lower = 0, upper = x)$value
  })
}

# Define the negative log-likelihood function with intercept fixed at -8
neg_log_likelihood_fixed_intercept <- function(params, times, event, manufacturer, warranty) {
  beta1 <- params[1]  # Warranty effect
  beta2 <- params[2]  # Manufacturer 2 effect
  beta3 <- params[3]  # Manufacturer 3 effect
  beta0 <- -8  # Fixed intercept

  # Compute the linear predictor (eta) including the intercept and covariates
  eta <- beta0 + beta1 * warranty + beta2 * (manufacturer == 2) + beta3 * (manufacturer == 3)

  # Calculate the log-likelihood for each individual
  loglik <- -sum(event * (eta + log(alpha_t(times))) - H0_t(times) * exp(eta))

  return(loglik)
}

# Prepare data
times <- data$days
event <- data$status
manufacturer <- as.numeric(data$manufacturer)
warranty <- data$warranty

# Fit the model using optimization with hessian=TRUE
fit <- optim(par = c(0, 0, 0),
             fn = neg_log_likelihood_fixed_intercept,
             times = times, event = event, manufacturer = manufacturer, warranty = warranty,
             hessian = TRUE)

# Extract parameter estimates and Hessian
params <- fit$par
hessian <- fit$hessian

# Calculate standard errors from the Hessian matrix
se <- sqrt(diag(solve(hessian)))

# Confidence interval for warranty effect (theta1)
conf_interval_theta1 <- params[1] + c(-1.96, 1.96) * se[1]
cat("Confidence Interval for Warranty Effect (theta1):", conf_interval_theta1, "\n")

# Confidence intervals for theta2 and theta3
conf_interval_theta2 <- params[2] + c(-1.96, 1.96) * se[2]
conf_interval_theta3 <- params[3] + c(-1.96, 1.96) * se[3]

# Print the confidence intervals
cat("Confidence Interval for Manufacturer 2 Effect (theta2):", conf_interval_theta2, "\n")
cat("Confidence Interval for Manufacturer 3 Effect (theta3):", conf_interval_theta3, "\n")

# Calculate 95% confidence region for (theta2, theta3)
theta2_range <- seq(params[2] - 3 * se[2], params[2] + 3 * se[2], length.out = 100)
theta3_range <- seq(params[3] - 3 * se[3], params[3] + 3 * se[3], length.out = 100)
grid <- expand.grid(theta2_range, theta3_range)

conf_region <- matrix(NA, nrow = nrow(grid), ncol = nrow(grid))

```
